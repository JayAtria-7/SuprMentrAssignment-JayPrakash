## 1. Problem Statement

Customer churn is a critical issue for telecommunications companies, leading to significant revenue loss. Acquiring a new customer is often much more expensive than retaining an existing one. The objective of this project is to build a predictive machine learning model that can identify customers who are at high risk of churning (canceling their subscriptions). By predicting churn proactively, the company can implement targeted retention strategies, improve customer satisfaction, and reduce revenue attrition.

## 2. Dataset

The proposed dataset for this project is the **Telco Customer Churn** dataset.

**Dataset Characteristics:**
- **Format:** CSV
- **Size:** ~7,000 instances, 21 attributes
- **Target Variable:** `Churn` (Categorical: Yes/No)

**Key Features:**
- *Demographics:* Gender, Age Range, Partners, Dependents.
- *Account Information:* Tenure, Contract type, Payment method, Monthly charges, Total charges.
- *Services Subscribed:* Phone service, Internet service type, Online security, Streaming.

## 3. Algorithm

This is a binary classification problem. We will evaluate and compare the following algorithms:

1. **Logistic Regression:** As a baseline model due to its interpretability.
2. **Random Forest Classifier:** An ensemble method handling non-linear relationships effectively.
3. **XGBoost (Extreme Gradient Boosting):** An advanced algorithm often yielding the highest accuracy for tabular data.

**Evaluation Metrics:**
We will prioritize metrics suitable for potentially imbalanced datasets:
- **Recall:** To minimize false negatives (missing a churner).
- **F1-Score:** The harmonic mean of precision and recall.
- **AUC-ROC:** To evaluate the overall model discrimination ability.

## 4. Expected Output

The expected deliverables of this mini-project include:

1. **Trained Machine Learning Model:** A robust classification model predicting churn probability.
2. **Exploratory Data Analysis (EDA) Report:** Visualizations highlighting key factors contributing to customer churn.
3. **Feature Importance Analysis:** Actionable business insights based on the most influential predictive features.
4. **Inference Script:** A script taking new customer data and outputting the likelihood of churn, allowing for proactive retention.

## 5. Work Description

This project will be executed in the following phases:
- **Phase 1: Data Preprocessing:** Handling missing values, encoding categorical variables, and scaling numerical features.
- **Phase 2: Exploratory Data Analysis (EDA):** Analyzing the distribution of variables to identify patterns and understand the differences between churners and non-churners.
- **Phase 3: Feature Engineering:** Creating new features or selecting the most relevant existing ones to improve predictive performance.
- **Phase 4: Model Training:** Splitting the dataset into train/test sets and training various classification models.
- **Phase 5: Model Evaluation & Tuning:** Testing combinations of hyperparameters, assessing model performance using selected metrics, and establishing the final model.
