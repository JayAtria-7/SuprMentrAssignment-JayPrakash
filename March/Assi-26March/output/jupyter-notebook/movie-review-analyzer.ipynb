# Movie Review Analyzer

**Goal:** Build a tiny, rule-based sentiment analyzer and test it on 5 movie reviews.

**Audience:** Beginners learning Python text processing.

**Prerequisites:** Basic Python (lists, functions, dictionaries).


## Outline
1. Setup a simple sentiment lexicon
2. Build a sentiment scoring function
3. Test on 5 reviews
4. Summarize results
5. Work description and learning outcome


```python
# Setup: minimal, deterministic imports
from __future__ import annotations
import re
```

## Step 1 - Define a tiny sentiment lexicon
We will use two small word lists: **positive** and **negative**. Each matched word adds or subtracts 1 point.


```python
positive_words = {
    "amazing", "great", "good", "excellent", "love", "loved", "enjoyed", "fantastic",
    "brilliant", "wonderful", "fun", "strong", "best", "beautiful", "uplifting"
}

negative_words = {
    "bad", "boring", "poor", "terrible", "awful", "hate", "hated", "disappointing",
    "weak", "worst", "dull", "mess", "slow", "predictable", "waste"
}

def tokenize(text: str) -> list[str]:
    'Lowercase and split into alphabetic tokens.'
    return re.findall(r"[a-z]+", text.lower())
```

## Step 2 - Build the sentiment analyzer
We count how many positive and negative words appear, then compute a score and label.


```python
def analyze_review(text: str) -> dict[str, object]:
    tokens = tokenize(text)
    pos = sum(1 for t in tokens if t in positive_words)
    neg = sum(1 for t in tokens if t in negative_words)
    score = pos - neg

    if score > 0:
        label = "Positive"
    elif score < 0:
        label = "Negative"
    else:
        label = "Neutral"

    return {
        "review": text,
        "positive": pos,
        "negative": neg,
        "score": score,
        "label": label,
    }
```

## Step 3 - Test on 5 reviews
These are short, mixed reviews to see how the rule-based model behaves.


```python
reviews = [
    "An amazing and uplifting film with great performances.",
    "The plot was boring and predictable, a complete waste of time.",
    "Good visuals but the story was slow and dull.",
    "I loved the soundtrack and the brilliant cinematography.",
    "Not the worst, but the acting was weak and the pacing was poor.",
]
```

## Step 4 - Run the analysis and summarize


```python
results = [analyze_review(r) for r in reviews]

# Pretty print a small table
print(f"{'Score':>5}  {'Label':>8}  Review")
print("-" * 60)
for r in results:
    print(f"{r['score']:>5}  {r['label']:>8}  {r['review']}")

# Summary counts
summary = {"Positive": 0, "Negative": 0, "Neutral": 0}
for r in results:
    summary[r["label"]] += 1

print("
Summary:")
for k, v in summary.items():
    print(f"{k}: {v}")
```
