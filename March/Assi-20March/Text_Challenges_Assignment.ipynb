## Part 2: Explanation of Preprocessing Needed

To effectively use these sentences in NLP models, we would need to apply several preprocessing steps to clean, normalize, and standardize the text data. Below is the proposed preprocessing pipeline for such noisy, informal text:

### 1. Lowercasing
Converting all text to lowercase helps reduce the vocabulary size and ensures that words like `Omw` and `omw` are treated as the exact same token by the model.

### 2. Emoji Handling (Demojization / Removal)
Emojis can carry significant sentiment, so sometimes outright removal isn't the best policy. 
- **Demojization**: We can use libraries like Python's `emoji` to convert them into text representations (e.g., `💀` becomes `:skull:`), which models can then interpret.
- **Removal**: If the task doesn't require sentiment analysis, emojis can simply be stripped out using regular expressions.

### 3. Spelling Correction (Typo Fixing)
Informal text contains numerous misspellings and elongated words (`everrr`, `ittt`). 
- **De-elongation**: We can use regular expressions to reduce repeated characters to a maximum of 2 or 1 character (e.g., `everrr` -> `ever`).
- **Spell-checking**: Applying context-aware spell checkers (like `pyspellchecker`, `SymSpell`, `TextBlob`, or large transformer models) to fix words like `exhousted` to `exhausted`.

### 4. Slang and Chat Word Expansion
Abbreviations and internet slang must be mapped back to their standard English equivalents so algorithms can understand them. 
- We would typically use a predefined slang dictionary lookup (e.g., `brb` -> `be right back`, `tbh` -> `to be honest`, `rn` -> `right now`).

### 5. Punctuation Handling
Messy text usually contains excessive punctuation (`??`, `!!!`, `...`). We usually normalize repeated punctuation into a single mark or simply remove all non-alphanumeric characters, depending on the model's needs.

### 6. Tokenization
After all the noise is cleaned up, text is split into tokens (words, sub-words, or characters) to convert the string of text into an array of items (e.g., `["on", "my", "way", "to", "the", "store"]`).

### 7. Lemmatization / Stemming
Finally, depending on the use case, words are reduced to their root forms (e.g., `running` -> `run`, `better` -> `good`), which further shrinks the vocabulary and improves exact matching.
