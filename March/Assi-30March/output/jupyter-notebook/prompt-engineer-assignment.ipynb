# Prompt Engineer Assignment


## Objective
Write clear prompts for three tasks and compare weak vs strong prompts. Finish with a concise work description and learning outcomes.


## Deliverables
1. Resume prompt (weak vs strong).
2. Business idea prompt (weak vs strong).
3. Study plan prompt (weak vs strong).
4. Short comparison of weak vs strong prompts.
5. Work description and learning outcomes.


## 1) Resume Prompt
**Weak Prompt**
```
Make a resume for me.
```
**Strong Prompt**
```
Role: You are a professional resume writer.
Task: Create a one-page resume for a software engineer.
Background: 3 years of experience, Python/JavaScript, built 2 SaaS products, led a team of 3, improved API latency by 40%.
Target: Mid-level backend engineer roles in US startups.
Constraints: ATS-friendly, reverse-chronological, no tables, 450?600 words.
Output format: Summary, Skills, Experience (3 bullets each role), Projects (2 bullets each), Education.
Tone: Confident and concise.
```
**Why stronger?** It defines role, context, audience, constraints, and output format.


## 2) Business Idea Prompt
**Weak Prompt**
```
Give me a business idea.
```
**Strong Prompt**
```
Role: You are a startup advisor.
Task: Propose 3 business ideas.
Context: I have $5,000 budget, 10 hours/week, strong skills in web development and digital marketing.
Industry focus: B2B productivity tools or creator economy.
Constraints: MVP in 6 weeks, low regulatory risk, remote-friendly.
Output format: For each idea include problem, target customer, solution, unique advantage, MVP scope, and first 3 validation steps.
```
**Why stronger?** It narrows the domain and forces actionable, structured outputs.


## 3) Study Plan Prompt
**Weak Prompt**
```
Make a study plan for me.
```
**Strong Prompt**
```
Role: You are an academic coach.
Task: Create a 6-week study plan to learn Data Structures and Algorithms.
Background: I can study 1.5 hours on weekdays and 4 hours on weekends.
Goal: Be ready for entry-level coding interviews.
Constraints: Use free resources, include weekly topics, daily tasks, and checkpoints.
Output format: Week-by-week table with topics, resources, and practice problems.
```
**Why stronger?** It defines time budget, goal, constraints, and required structure.


## 4) Weak vs Strong Prompt Comparison
Strong prompts typically include the following elements.
- Role or perspective for the model.
- Clear task description and objective.
- User context and constraints.
- Audience or target use case.
- Output format and length guidance.
- Success criteria or checkpoints.
