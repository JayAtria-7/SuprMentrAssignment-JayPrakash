# Assignment: Build a Text Cleaner
**Description:** Write code to remove punctuation, lowercase text, remove stopwords and test it.

### Work Description
This project involves developing a modular Python function to perform fundamental text preprocessing tasks, an essential initial step in Natural Language Processing (NLP). The text cleaner takes raw, messy strings of text and applies a three-step transformation pipeline: First, it converts all characters to lowercase to ensure uniformity. Second, it strips out all punctuation marks using Python's built-in string manipulation tables. Finally, it tokenizes the text and removes common English "stopwords" (such as "is", "the", "and") by filtering the words against the Natural Language Toolkit (NLTK) corpus, leaving behind only the most meaningful words for subsequent analysis.

### Learning Outcomes
- **Text Normalization:** Understand how to standardize text data by converting it into a uniform case.
- **Data Cleansing:** Learn efficient string manipulation techniques in Python, specifically how to use translation tables (`str.maketrans()`) to strip out unneeded characters like punctuation marks.
- **Stopword Removal:** Gain practical experience with NLTK by implementing custom filtering logic to eliminate high-frequency, low-meaning words.
- **Modular Coding & Testing:** Practice writing reusable code structures (functions) and testing them against multiple sample sentences to verify expected behavior.

```python
# Install nltk if not already installed (uncomment the line below if needed)
# !pip install nltk

import string
import nltk

# Download the stopwords dataset
nltk.download('stopwords', quiet=True)
from nltk.corpus import stopwords
```

```python
def clean_text(text):
    """
    Cleans the input text by:
    1. Lowercasing the text
    2. Removing punctuation
    3. Removing stopwords
    """
    # 1. Lowercase text
    text = text.lower()
    
    # 2. Remove punctuation
    # Create a translation table that maps every punctuation character to None
    translator = str.maketrans('', '', string.punctuation)
    text = text.translate(translator)
    
    # 3. Remove stopwords
    stop_words = set(stopwords.words('english'))
    words = text.split()
    cleaned_words = [word for word in words if word not in stop_words]
    
    # Rejoin the words into a single string
    return " ".join(cleaned_words)
```

```python
# Test cases
test_sentences = [
    "Hello World! This is a simple test.",
    "Data Science is AWESOME!!! Don't you think?",
    "Wait, what? I thought we were going to the park...",
    "NLTK makes text processing VERY easy and fast."
]

for i, sentence in enumerate(test_sentences):
    print(f"--- Test Case {i+1} ---")
    print(f"Original: {sentence}")
    print(f"Cleaned:  {clean_text(sentence)}\n")
```
